```python
from bs4 import BeautifulSoup


html=BeautifulSoup(result.content)
```

```python
html
```

```python
quotes=html.find("span")
quotes
```

```python
quotes=html.find_all("span",class_="text")
quotes
```

```python
qu=[]

for i in quotes:
    qu.append(i.text)
qu
```

```python
author=html.find("small",html.parser)
author
```

```python
author=html.find("small","lxml")
author
```

```python
author=html.find_all("small")
```

```python
author
```

```python
au=[]
for i in author:
    au.append(i.text)
```

```python
au
```

```python
qu
```

```python
tags=html.find_all("div",class_="tags")
tags

tag=[]
for i in range(len(tags)):
    diff=[]
    for j in tags[i].find_all("a",class_="tag"):
        diff.append(j.text)
    tag.append(diff)
tag
```

```python
import pandas as pd

df=pd.DataFrame()
df["Author"]=au
df["QUOTES"]=qu
df["KEYWORDS"]=tag
```

```python
df
```
